```python
# Read the PDF 
reader = PdfReader(pdf_path)

print("Total PDF pages:", len(reader.pages))
```

```python
pages = []

for pdf_page_number, page in enumerate(
    tqdm(reader.pages),
    start=1
):
    
    text = page.extract_text()
    
    if text:
        pages.append({
            "pdf_page": pdf_page_number,
            "text": text
        })

print("Pages containing text:", len(pages))
```

```python
#Text Cleaning Function 
def clean_text(text):
    
    # Replace PDF soft-hyphen / unusual hyphen characters
    text = text.replace("\u00ad", "")
    text = text.replace("\ufffd", "")
    
    # Fix common PDF hyphenation across line breaks
    text = re.sub(
        r"(\w)[-–]\s*\n\s*(\w)",
        r"\1\2",
        text
    )
    
    # Replace multiple spaces
    text = re.sub(r"[ \t]+", " ", text)
    
    # Replace multiple newlines
    text = re.sub(r"\n+", "\n", text)
    
    # Remove spaces before punctuation
    text = re.sub(r"\s+([,.!?;:])", r"\1", text)
    
    return text.strip()
```

```python
# Cleaning The Pages 
cleaned_pages = []

for page in tqdm(pages):
    
    cleaned = clean_text(page["text"])
    
    if len(cleaned) > 50:
        cleaned_pages.append({
            "pdf_page": page["pdf_page"],
            "text": cleaned
        })

print("Cleaned pages:", len(cleaned_pages))
```

```python
# Detect chapter pages
def normalize_heading(text):
    text = text.upper()
    text = text.replace("—", "-")
    text = text.replace("–", "-")
    text = re.sub(r"[^A-Z0-9]+", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
# Creating the RAG chunks
def create_chunks(
    text,
    chunk_size=800,
    overlap=100
):
    
    words = text.split()
    
    chunks = []
    
    start = 0
    
    while start < len(words):
        
        end = min(
            start + chunk_size,
            len(words)
        )
        
        chunk = " ".join(
            words[start:end]
        )
        
        if len(chunk.strip()) > 50:
            chunks.append(chunk.strip())
        
        if end >= len(words):
            break
        
        start = end - overlap
    
    return chunks
```

```python
# Generating the corpus
corpus = []

chunk_id = 0

for chapter_range in chapter_ranges:
    
    chapter_records = [
        record
        for record in chapter_pages
        if record["chapter_number"]
        == chapter_range["chapter_number"]
    ]
    
    # Combine pages belonging to the chapter
    combined_text = " ".join(
        record["text"]
        for record in chapter_records
    )
    
    chunks = create_chunks(
        combined_text
    )
    
    for chunk_number, chunk in enumerate(
        chunks,
        start=1
    ):
        
        corpus.append({
            "id": f"chunk_{chunk_id}",
            "chapter_number":
                chapter_range["chapter_number"],
            "chapter":
                chapter_range["chapter"],
            "chunk_number":
                chunk_number,
            "text":
                chunk
        })
        
        chunk_id += 1

print("Total chunks:", len(corpus))
```

Better Text Cleaning 

```python
def clean_ncert_text(text):

    # Remove year/footer
    text = re.sub(r"\b2018[-–]19\b", " ", text)

    # Remove replacement characters
    text = text.replace("\ufffd", " ")

    # Remove soft hyphens
    text = text.replace("\u00ad", "")

    # Join words broken by line/hyphen
    text = re.sub(
        r"(\w)[-–]\s+(\w)",
        r"\1\2",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()
```

```python
# Applying Text Cleaning 
for record in chapter_pages:
    record["text"] = clean_ncert_text(
        record["text"]
    )

print("Text cleaning completed.")
```

Creating smaller RAG chunks

I want to change our previous chunking approach.

Instead of 800 words, let's use approximately 350 words with 50-word overlap.

This is better for our RAG system because the retrieved context will be more focused.

```python
def create_chunks(
    words,
    chunk_size=350,
    overlap=50
):

    chunks = []

    start = 0

    while start < len(words):

        end = min(
            start + chunk_size,
            len(words)
        )

        chunk_words = words[start:end]

        if len(chunk_words) >= 40:
            chunks.append(
                (start, end, chunk_words)
            )

        if end >= len(words):
            break

        start = end - overlap

    return chunks
```

Building the Final Corpus 

```python
corpus = []

chunk_id = 0

for chapter in chapter_ranges:

    records = [
        record
        for record in chapter_pages
        if record["chapter_number"]
        == chapter["chapter_number"]
    ]

    # Sort pages
    records = sorted(
        records,
        key=lambda x: x["pdf_page"]
    )

    # Process each page separately first
    # so we don't lose page information

    for record in records:

        words = record["text"].split()

        chunks = create_chunks(words)

        for chunk_number, (_, _, chunk_words) in enumerate(
            chunks,
            start=1
        ):

            chunk_text = " ".join(chunk_words)

            corpus.append({
                "id": f"chunk_{chunk_id}",
                "chapter_number": chapter["chapter_number"],
                "chapter": chapter["chapter"],
                "chunk_number": chunk_number,
                "pdf_page": record["pdf_page"],
                "printed_page": record["printed_page"],
                "text": chunk_text
            })

            chunk_id += 1

print("Total chunks:", len(corpus))
```

```python
import json
import os

import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer
import faiss

from haystack import Document
```

```python
# Load our JSONL Corpus 
corpus_file = "class8_science.jsonl"

documents = []

with open(corpus_file, "r", encoding="utf-8") as f:
    for line in f:
        documents.append(json.loads(line))

print("Total documents:", len(documents))
```

```python
# Load the Embedding Model 

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

```python
# Create a Semantic Search Function 
def semantic_search(
    query,
    top_k=5
):

    # Convert question to embedding
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    # Search FAISS
    scores, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        result = documents[idx].copy()

        result["similarity_score"] = float(score)

        results.append(result)

    return results
```

```python
# Loading the FLAN-T5-Small 

MODEL_NAME = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

llm = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME
)

print("FLAN-T5-Small loaded successfully.")
```

```python
# Select CPU/GPU automatically 
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

llm = llm.to(device)

print("Using device:", device)
```

```python
# Creating the RAG Prompt 
def create_prompt(question, retrieved_documents):

    context_parts = []

    # Use only the top 3 retrieved chunks
    for i, document in enumerate(
        retrieved_documents[:3],
        start=1
    ):

        # Limit each chunk
        words = document["text"].split()
        limited_text = " ".join(words[:180])

        context_parts.append(
            f"SOURCE {i}\n"
            f"Chapter: {document['chapter']}\n"
            f"Page: {document['printed_page']}\n"
            f"Text: {limited_text}\n"
        )

    context = "\n".join(context_parts)

    prompt = f"""
You are an AI tutor for NCERT Class 8 Science.

Answer the question using ONLY the textbook context below.

Rules:
- Do not use outside knowledge.
- Do not make up information.
- Use simple language suitable for a Class 8 student.
- If the context does not contain enough information, say:
  "I'm focused on Class 8 Science. I could not find this information in the available textbook material."

QUESTION:
{question}

TEXTBOOK CONTEXT:
{context}

FINAL ANSWER:
"""

    return prompt
```

```python
# Creating the LLM Generation Function 
def generate_answer(prompt):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        output = llm.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=4,
            no_repeat_ngram_size=2
        )

    answer = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return answer.strip()
```

```python
# Build the complete RAG function
# Now we're combining: FAISS retrieval, context construction, LLM generation, source citations
def rag_answer(question, top_k=3):

    # Retrieve relevant textbook chunks
    retrieved_documents = semantic_search(
        question,
        top_k=top_k
    )

    # Check whether retrieval found relevant material
    if not retrieved_documents:
        return {
            "question": question,
            "answer": (
                "I'm focused on Class 8 Science. "
                "I could not find this information "
                "in the available textbook material."
            ),
            "sources": [],
            "retrieved_documents": []
        }

    # Get best similarity score
    best_score = retrieved_documents[0][
        "similarity_score"
    ]

    # Out-of-syllabus threshold
    if best_score < 0.35:

        return {
            "question": question,
            "answer": (
                "I'm focused on Class 8 Science. "
                "I could not find this information "
                "in the available textbook material."
            ),
            "sources": [],
            "retrieved_documents": []
        }

    # Create prompt
    prompt = create_prompt(
        question,
        retrieved_documents
    )

    # Generate answer
    answer = generate_answer(prompt)

    # Prepare sources
    sources = []

    for document in retrieved_documents:

        sources.append({
            "chapter": document["chapter"],
            "page": document["printed_page"],
            "similarity": round(
                document["similarity_score"],
                4
            )
        })

    return {
        "question": question,
        "answer": answer,
        "sources": sources,
        "retrieved_documents": retrieved_documents
    }
```

```python
def is_in_scope(retrieved_documents):

    if not retrieved_documents:
        return False

    best_score = retrieved_documents[0][
        "similarity_score"
    ]

    return best_score >= SIMILARITY_THRESHOLD
```

```python
# Updating the RAG Function 
def rag_answer(
    question,
    top_k=5
):

    # Retrieve textbook chunks
    retrieved_documents = semantic_search(
        question,
        top_k=top_k
    )

    # Check whether question appears
    # to be within the textbook domain
    if not is_in_scope(
        retrieved_documents
    ):

        return {
            "question": question,
            "answer": (
                "I'm focused on Class 8 Science. "
                "I could not find this information "
                "in the available textbook material."
            ),
            "sources": [],
            "retrieved_documents": []
        }

    # Create RAG prompt
    prompt = create_prompt(
        question,
        retrieved_documents
    )

    # Generate answer
    answer = generate_answer(
        prompt
    )

    # Prepare citations
    sources = []

    for document in retrieved_documents:

        sources.append({
            "chapter": document["chapter"],
            "page": document["printed_page"],
            "similarity": round(
                document["similarity_score"],
                4
            )
        })

    return {
        "question": question,
        "answer": answer,
        "sources": sources,
        "retrieved_documents":
            retrieved_documents
    }
```

```python
# Creating the BLEU Function 
smoothie = SmoothingFunction().method4


def calculate_bleu(
    reference,
    candidate
):

    reference_tokens = reference.lower().split()
    candidate_tokens = candidate.lower().split()

    score = sentence_bleu(
        [reference_tokens],
        candidate_tokens,
        smoothing_function=smoothie
    )

    return score
```

```python
# Creating the ROUGE-L Function 
rouge = rouge_scorer.RougeScorer(
    ["rougeL"],
    use_stemmer=True
)


def calculate_rouge_l(
    reference,
    candidate
):

    score = rouge.score(
        reference,
        candidate
    )

    return score["rougeL"].fmeasure
```

```python
# Calculating BLEU and ROUGE-L
for result in evaluation_results:

    result["BLEU"] = calculate_bleu(
        result["reference_answer"],
        result["generated_answer"]
    )

    result["ROUGE-L"] = calculate_rouge_l(
        result["reference_answer"],
        result["generated_answer"]
    )
```

```python
# Create Logging Function 
from datetime import datetime

def log_interaction(
    question,
    result
):

    # Get best similarity score
    if result["sources"]:

        best_similarity = max(
            source["similarity"]
            for source in result["sources"]
        )

        source_chapters = "; ".join(
            dict.fromkeys(
                source["chapter"]
                for source in result["sources"]
            )
        )

        source_pages = "; ".join(
            str(source["page"])
            for source in result["sources"]
        )

    else:

        best_similarity = None
        source_chapters = ""
        source_pages = ""

    interaction_logs.append({

        "timestamp": datetime.now().isoformat(
            timespec="seconds"
        ),

        "question": question,

        "answer": result["answer"],

        "best_similarity": best_similarity,

        "source_chapters": source_chapters,

        "source_pages": source_pages,

        "num_sources": len(
            result["sources"]
        )
    })


print("Logging function created.")
```

```python
# Test Interaction Logging 
question = "What is friction?"

result = rag_answer(
    question,
    top_k=3
)

log_interaction(
    question,
    result
)

print(
    "Interactions logged:",
    len(interaction_logs)
)
```

```python
def classify_error(row):

    # Check if human review has not been completed
    if pd.isna(row["Correctness"]) or pd.isna(row["Relevance"]):
        return "Not Reviewed"

    # Human evaluation
    if row["Correctness"] <= 2:
        return "Incorrect Answer"

    elif row["Relevance"] <= 2:
        return "Irrelevant Answer"

    # Automatic metrics
    elif row["BLEU"] < 0.01:
        return "Low Lexical Overlap"

    elif row["ROUGE-L"] < 0.10:
        return "Low ROUGE-L"

    else:
        return "Acceptable"


analysis_df["Error_Type"] = analysis_df.apply(
    classify_error,
    axis=1
)

display(
    analysis_df[
        [
            "question",
            "BLEU",
            "ROUGE-L",
            "Correctness",
            "Relevance",
            "Error_Type"
        ]
    ]
)
```

```python
# Error type summary 
error_summary = (
    analysis_df[
        "Error_Type"
    ]
    .value_counts()
    .reset_index()
)

error_summary.columns = [
    "Error_Type",
    "Count"
]

display(error_summary)
```
